## Assignment 5.3
**Question:** WAP to add all prime numbers from 1 to n.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

n = int(input("Enter n: "))
total = 0
for i in range(2, n+1):
    if is_prime(i):
        total += i
print("Sum =", total)
```

## Assignment 6.2
**Question:** WAP using a function to add all prime numbers from 1 to n.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

def prime_sum(n):
    total = 0
    for i in range(2, n+1):
        if is_prime(i):
            total += i
    return total

n = int(input("Enter n: "))
print("Sum =", prime_sum(n))
```
