## Q3. Random Numbers

Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate). 

i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

ii. Count and print all odd numbers in the list. 

iii. Count and print all even numbers in the list. 

iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension. 

v. Print the number that occurs most frequently in your list, and how many times it occurs. 

```python
#i)
import random

random.seed(12345678)
nums = [random.randint(100,900) for _ in range(100)]

odd = [n for n in nums if n%2]
even = [n for n in nums if n%2==0]

def prime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

primes = [n for n in nums if prime(n)]

from collections import Counter
c = Counter(nums)
num,freq = c.most_common(1)[0]

#ii)
print("ii. Odd count:", len(odd))
#iii)
print("iii. Even count:", len(even))
#iv)
print("iv. Prime count:", len(primes))
print("Prime list:", primes)
#v)
print("v. Most frequent:", num, "Frequency:", freq)
```

## Q4. Sets

 Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets. 
 
 vi. Find the union of A and B — all unique values across both sets. 
 
 vii. Find the intersection of A and B — values common to both. 
 
 viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference(). 
 
 ix. Find the symmetric difference of A and B. 
 x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset(). 
 xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists. 

```python
digits = [int(d) for d in "10241704"]
A = {d*7 for d in digits}
B = {d*9 for d in digits}

print("A:",A)
print("B:",B)
#vi)
print("Union:", A|B)
#vii)
print("Intersection:", A&B)
#viii)
print("A-B:", A.difference(B))
print("B-B:", B.difference(A))
# difference() returns elements only in one set, symmetric_difference() returns elements exclusive to both.

#ix)
print("Symmetric Difference:", A.symmetric_difference(B))
#x)
print("Subset:", A.issubset(B))
print("Superset:", B.issuperset(A))
#xi)
x = int(input("Enter X: "))
A.discard(x)
print("Updated A:",A)
# discard() does not raise an error if the value is absent.
```

```python
my_dict = {
    "name":"Parth Khurana",
    "roll_no":"1024170354",
    "branch":"CSE",
    "age":20,
    "city":"Mumbai"
}

#i)
my_dict["location"] = my_dict.pop("city")
print("i.", my_dict)

#ii)
my_dict["cgpa"] = 8.91
print("ii.", my_dict)

#iii)
my_dict["age"] += 1
print("iii.", my_dict)

#iv)
d1 = my_dict.copy()
removed = d1.pop("branch")
print("iv. pop:", d1, "Removed:", removed)

d2 = my_dict.copy()
del d2["branch"]
print("del:", d2)
# pop() returns removed value; del only deletes.

#v)
for k,v in my_dict.items():
    print(k,"→",v)

#vi)
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("vi. Email key not found.")

#vii)
friend_dict = {
    "name":"Rahul",
    "roll_no":"1024567876",
    "branch":"CSE",
    "age":21,
    "city":"Delhi"
}

merged = {**my_dict, **friend_dict}
print("vii.", merged)
# When keys overlap, values from the second dictionary win.

#viii)
strings = {k:v for k,v in my_dict.items() if isinstance(v,str)}
print("viii.", strings)
```
