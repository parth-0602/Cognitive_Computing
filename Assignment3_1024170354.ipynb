## Q1 — Building the required DataFrame

First, I have entered the values from the question into Python lists and then combined them into one Pandas DataFrame. Keeping the data in a dictionary makes the column structure easy to read.


```python
records = [
    [1, "Yes", "Single", "125K", "No"],
    [2, "No", "Married", "100K", "No"],
    [3, "No", "Single", "70K", "No"],
    [4, "Yes", "Married", "120K", "No"],
    [5, "No", "Divorced", "95K", "Yes"],
    [6, "No", "Married", "60K", "No"],
    [7, "Yes", "Divorced", "220K", "No"],
    [8, "No", "Single", "85K", "Yes"],
    [9, "No", "Married", "75K", "No"],
    [10, "No", "Single", "90K", "Yes"]
]

columns = ["Tid", "Refund", "Marital Status", "Taxable Income", "Cheat"]
df = pd.DataFrame(records, columns=columns)

display(df)
```

## Q2 — Selecting specific rows

The required rows are selected with `loc`. Since the DataFrame keeps the default integer labels, the requested labels can be passed together as a list.


```python
selected_rows = df.loc[[0, 4, 7, 8]]
display(selected_rows)
```

## Q3 — Working with `iloc`

Here I use positional indexing. The ending position in an `iloc` slice is not included, so the stop values are one position beyond the last row or column wanted.


```python
print("1) Rows 3 through 7")
display(df.iloc[3:8])

print("2) Rows 4 through 8, columns 2 through 4")
display(df.iloc[4:9, 2:5])

print("3) Every row, columns 1 through 3")
display(df.iloc[:, 1:4])
```

## Q4 — Loading and previewing the Iris CSV

The Iris dataset is read with `read_csv()` and the first five records are displayed. If the CSV is missing, the notebook tries to obtain the Kaggle copy automatically.


```python
from pathlib import Path
import urllib.request
import zipfile

iris_path = Path("Iris.csv")

if not iris_path.is_file():
    download_url = "https://www.kaggle.com/api/v1/datasets/download/uciml/iris"
    archive = Path("iris_dataset.zip")

    try:
        urllib.request.urlretrieve(download_url, archive)
        with zipfile.ZipFile(archive) as zf:
            zf.extractall(".")
        print("Iris CSV downloaded.")
    except Exception as err:
        print("Automatic download failed:", err)

iris_df = pd.read_csv(iris_path)
print("First five records:")
display(iris_df.iloc[:5])
```

## Q5 — Removing a row and a column

For this question, row 4 and column 3 are treated as zero-based positions. The row is removed first, followed by the fourth column.


```python
iris_modified = iris_df.drop(index=4).copy()

removed_column = iris_modified.columns[3]
iris_modified.drop(columns=removed_column, inplace=True)

print("Removed row:", 4)
print("Removed column:", removed_column)
display(iris_modified.head(10))
```

# Q6 — Employee DataFrame

The employee information from the question is stored in a dictionary, converted into a DataFrame, and also written to `employees.csv` for the later operations.


```python
employee_data = [
    [101, "Alice", "HR", 29, 50000, 4, "2020-03-15", "Female", 5000, 4.5],
    [102, "Bob", "IT", 34, 70000, 8, "2017-07-19", "Male", 7000, 4.0],
    [103, "Charlie", "IT", 41, 65000, 10, "2013-06-01", "Male", 6000, 3.8],
    [104, "Diana", "Marketing", 28, 55000, 3, "2021-02-10", "Female", 4500, 4.7],
    [105, "Edward", "Sales", 38, 60000, 12, "2010-11-25", "Male", 5000, 3.5]
]

employee_columns = [
    "Employee_ID", "Name", "Department", "Age", "Salary",
    "Years_of_Experience", "Joining_Date", "Gender", "Bonus", "Rating"
]

employees_df = pd.DataFrame(employee_data, columns=employee_columns)
employees_df["Joining_Date"] = pd.to_datetime(employees_df["Joining_Date"])

employees_df.to_csv("employees.csv", index=False)

print("employees.csv has been created.")
display(employees_df)
```

## Q6(a) — DataFrame dimensions

The `shape` attribute reports the number of rows followed by the number of columns.


```python
rows, columns = employees_df.shape
print(f"Rows: {rows}")
print(f"Columns: {columns}")
```

## Q6(b) — Inspecting the DataFrame

`info()` is useful for checking column names, data types, and whether any values are missing.


```python
employees_df.info()
```

## Q6(c) — Numerical summary

For numerical fields, `describe()` provides the main descriptive statistics such as mean, quartiles, minimum, and maximum.

## Q6(e) — Calculating the requested values

The following expressions calculate the average salary, total bonus, minimum age, and maximum rating directly from their respective columns.


```python
avg_salary = employees_df["Salary"].mean()
bonus_total = employees_df["Bonus"].sum()
min_age = employees_df["Age"].min()
top_rating = employees_df["Rating"].max()

print(f"1. Average salary: ₹{avg_salary:,.2f}")
print(f"2. Total bonus: ₹{bonus_total:,.2f}")
print(f"3. Youngest age: {min_age}")
print(f"4. Highest rating: {top_rating}")
```

## Q6(f) — Salary ranking

Sorting the salary column in descending order places the highest-paid employee at the top.


```python
salary_order = employees_df.sort_values(
    "Salary",
    ascending=False
)
display(salary_order)
```

## Q6(g) — Converting ratings into categories

A small function maps each rating to the category specified in the question. The result is stored in a new column.


```python
def classify_rating(value):
    if value >= 4.5:
        return "Excellent"
    if value >= 4.0:
        return "Good"
    return "Average"

employees_df["Performance_Category"] = employees_df["Rating"].map(classify_rating)
display(employees_df)
```

## Q6(h) — Checking for missing entries

`isnull().sum()` gives a column-wise count of missing values.


```python
null_counts = employees_df.isna().sum()
print("Missing values by column:")
display(null_counts)
```

## Q6(i) — Renaming the employee identifier

The `Employee_ID` column is renamed to the shorter label `ID`.


```python
employees_df.rename(
    columns={"Employee_ID": "ID"},
    inplace=True
)
display(employees_df)
```

## Q6(j) — Filtering employee records

Two filters are applied separately: one for experience greater than five years and another for employees in the IT department.


```python
experienced = employees_df.query("Years_of_Experience > 5")
it_team = employees_df.loc[employees_df["Department"].eq("IT")]

print("Employees with more than 5 years of experience:")
display(experienced)

print("Employees working in IT:")
display(it_team)
```

## Q6(k) — Computing employee tax

The tax column is calculated from salary using the 10% rate given in the question.


```python
employees_df["Tax"] = employees_df["Salary"].mul(0.10)
display(employees_df)
```

## Q6(l) — Exporting the final DataFrame

Finally, the updated DataFrame is saved as `employees_modified.csv` and read back once to confirm the saved file.


```python
save_path = "employees_modified.csv"
employees_df.to_csv(save_path, index=False)

print(f"Saved updated employee data to {save_path}")
display(pd.read_csv(save_path))
```
