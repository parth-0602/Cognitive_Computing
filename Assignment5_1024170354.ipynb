## Q1 — Basic operations on a 1-D array

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)
```

## Q2 — Reverse an array and find the most frequent values

```python
# Q2(a): Reverse the given array
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original array:", arr)
print("Reversed array:", arr[::-1])

# Q2(b)(i): Most frequent value and its indices in x
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_frequent_x = values[counts == max_count]
print("\nx:", x)
print("Most frequent value(s):", most_frequent_x)
for value in most_frequent_x:
    print(f"Indices of {value}:", np.where(x == value)[0])

# Q2(b)(ii): Most frequent value and its indices in y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
max_count = counts.max()
most_frequent_y = values[counts == max_count]
print("\ny:", y)
print("Most frequent value(s):", most_frequent_y)
for value in most_frequent_y:
    print(f"Indices of {value}:", np.where(y == value)[0])
```

## Q3 — Access elements in a 2-D array

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Array:\n", arr)
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

## Q4 — Create and inspect a 1-D array using linspace()

```python
parth_khurana = np.linspace(10, 100, 25)

print("Array:", parth_khurana)
print("Dimensions:", parth_khurana.ndim)
print("Shape:", parth_khurana.shape)
print("Total elements:", parth_khurana.size)
print("Data type:", parth_khurana.dtype)
print("Total bytes consumed:", parth_khurana.nbytes)

# Reshape to a 5 x 5 array to show a 2-D transpose
reshaped_array = parth_khurana.reshape(5, 5)
print("\nReshaped array (5 x 5):\n", reshaped_array)
print("Transpose using transpose():\n", reshaped_array.transpose())
print("Transpose using .T:\n", reshaped_array.T)
print("\nA 1-D array's .T does not change its shape; reshape it to 2-D first to see a matrix transpose.")
```

## Q5 — Array statistics, reshape(), and resize()

```python
ucs420_parth_khurana = np.array([[10, 20, 30, 40],
                                  [50, 60, 70, 80],
                                  [90, 15, 20, 35]])

print("Original array:\n", ucs420_parth_khurana)
print("Mean:", np.mean(ucs420_parth_khurana))
print("Median:", np.median(ucs420_parth_khurana))
print("Maximum:", np.max(ucs420_parth_khurana))
print("Minimum:", np.min(ucs420_parth_khurana))
print("Unique elements:", np.unique(ucs420_parth_khurana))

reshaped_ucs420_parth_khurana = ucs420_parth_khurana.reshape(4, 3)
print("\nReshaped array (4 x 3):\n", reshaped_ucs420_parth_khurana)

resized_ucs420_parth_khurana = np.resize(ucs420_parth_khurana, (2, 3))
print("\nResized array (2 x 3):\n", resized_ucs420_parth_khurana)
```
