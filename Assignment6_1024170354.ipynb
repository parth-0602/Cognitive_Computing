## Q1. Sensor readings

```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected = temperature + 2
print("a. Corrected Celsius:", corrected)
fahrenheit = corrected * 9/5 + 32
print("b. Fahrenheit:", fahrenheit)
above32 = corrected[corrected > 32]
print("c. Readings above 32°C:", above32)
print("d. Count above 32°C:", np.count_nonzero(corrected > 32))
```

## Q2. Daily steps

```python
steps = np.array([[5000,6200,7100],[8000,7500,9000],[4500,5100,4800],[9000,8500,9500]])
print("a. Total:", np.sum(steps))
print("b. Mean:", np.mean(steps))
print("c. Maximum:", np.max(steps), "Minimum:", np.min(steps))
print("d. Total each day (axis=0):", np.sum(steps, axis=0))
print("e. Total each user (axis=1):", np.sum(steps, axis=1))
pos = np.unravel_index(np.argmax(steps), steps.shape)
print("f. Maximum position (zero-based user, day):", pos, "value:", steps[pos])
print("   One-based user/day:", (pos[0]+1, pos[1]+1))
```

## Q3. Slicing, views, copies, reshape, flatten and ravel

```python
original = np.array([1,2,3,4,5,6])
subset = original[1:5]
subset[0] = 999
print("c. original after changing slice:", original, "subset:", subset)
copied = original[1:5].copy()
copied[0] = 500
print("d. original after changing copy:", original, "copy:", copied)
matrix = np.arange(1,13).reshape(3,4)
print("e. 3x4 matrix:\n", matrix)
print("f. First row:", matrix[0,:])
print("   Last row:", matrix[-1,:])
print("   Second column:", matrix[:,1])
print("   Rows 1-2 and columns 2-3 (one-based):\n", matrix[0:2,1:3])
flat = matrix.flatten()
ravelled = matrix.ravel()
print("g. flatten:", flat, "ravel:", ravelled)
ravelled[0] = 100
print("h. After changing ravel, matrix:\n", matrix)
flat[1] = 200
print("i. After changing flatten, matrix remains independent:\n", matrix)
print("   flattened copy:", flat)
print("j. shape:", matrix.shape, "ndim:", matrix.ndim, "size:", matrix.size, "dtype:", matrix.dtype)
```

## Q4. OLS assistance score

```python
X = np.array([[6,70,3],[5,50,6],[8,80,2],[4,30,8]], dtype=float)
y = np.array([40,65,30,85], dtype=float)
print("a. shape:", X.shape, "dimensions:", X.ndim)
XT = X.T
print("b. X transpose:\n", XT)
XTX = XT @ X
print("c. X.T @ X:\n", XTX)
print("d. Inverse of X.T @ X (if invertible):")
try:
    inverse = np.linalg.inv(XTX)
    print(inverse)
    beta = inverse @ XT @ y
except np.linalg.LinAlgError:
    print("Matrix singular; using pseudoinverse.")
    beta = np.linalg.pinv(XTX) @ XT @ y
print("e. OLS coefficients:", beta)
print("f. Coefficients correspond to sleep hours, activity level, stress level.")
print("   They estimate the linear change in predicted score per unit feature, holding others fixed.")
new_user = np.array([5,40,7], dtype=float)
print("g. Predicted score:", new_user @ beta)
```
